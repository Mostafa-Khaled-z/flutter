```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split

# 1. Load the dataset into a pandas DataFrame
data = {
    "AD_Budget": [1000, 2000, 3000, 4000, 5000, 6000],
    "Sales": [40, 65, 80, 100, 120, 150],
}
df = pd.DataFrame(data)

# 2. Define x (independent variable) and y (dependent variable)
# Note: scikit-learn expects X to be a 2D array/DataFrame
X = df[["AD_Budget"]]
y = df["Sales"]

# 3. Split the data into training and testing sets
# test_size = 0.4 means 40% of data goes to testing, 60% to training
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.4, random_state=42
)

# 4. Create a LinearRegression model
model = LinearRegression()

# 5. Fit the model to the training data
model.fit(X_train, y_train)

# 6. Print the coefficient and intercept of the regression line
a = model.coef_[0]
b = model.intercept_
print("--- Model Parameters ---")
print(f"Coefficient (a): {a:.4f}")
print(f"Intercept (b): {b:.4f}\n")

# 7. Predict the values of the test set and store them in y_pred
y_pred = model.predict(X_test)

# 8. Calculate the residuals (y_test - y_pred) and print them
residuals = y_test - y_pred
print("--- Residuals ---")
print(residuals.to_string(index=False))
print()

# 9. Compute and print the R2 Score and the Mean Squared Error (MSE)
r2 = r2_score(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
print("--- Evaluation Metrics ---")
print(f"R2 Score: {r2:.4f}")
print(f"Mean Squared Error (MSE): {mse:.4f}\n")

# 10. Make three new predictions for 2500, 4500, and 5500
# Note: The prompt text has a minor typo switching "AD_Budget" to "study hours/exam score",
# but we use the requested numerical values (2500, 4500, 5500) for our AD_Budget.
new_budgets = pd.DataFrame({"AD_Budget": [2500, 4500, 5500]})
new_predictions = model.predict(new_budgets)

print("--- New Predictions ---")
for budget, pred in zip([2500, 4500, 5500], new_predictions):
    print(f"Predicted Sales for AD_Budget ${budget}: {pred:.2f} units")

print("\n--- Final Regression Equation ---")
print(f"Sales = {a:.4f} * AD_Budget + {b:.4f}")
```
